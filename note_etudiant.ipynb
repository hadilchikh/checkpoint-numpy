```python
import numpy as np 
```

```python
#2.nombre d'étudiants et de matières
nb_etudiants=int(input("donner le nombre d'étudiants:"))
nb_matieres=int(input("donner le nombre de matières:"))
```

```python
#3.création d'un tableau numpy 
note = np.zeros((nb_etudiants, nb_matieres), dtype=float)
for i in range(nb_etudiants):
    for j in range(nb_matieres):
        note[i][j]=float(input("donner la note de l'étudiant {} pour la matière {}:".format(i+1,j+1)))
        while note[i][j] < 0 or note[i][j] > 100:
            print("Note invalide ! La note doit être entre 0 et 100.")
            note[i][j] = float(input("Entrez une autre note : "))
```

```python
#4.les notes totales
note
total=note.sum(axis=1)
total
```

```python
#5.le pourcentage de chque étudiant 
pourcentage=total/(nb_matieres*100)*100
```

```python
#6.la mention de chaque étudiant
mention=[]
for p in pourcentage:
    if p >= 90:
        mention.append("A+")
    elif p >= 80:
        mention.append("A")
    elif p >= 70:
        mention.append("B+")
    elif p >= 60:
        mention.append("B")
    elif p >= 50:
        mention.append("C")
    else:
        mention.append("F")
```

```python
print("\nRésultats des étudiants :")

for i in range(nb_etudiants):
    print("-------------------------")
    print("Étudiant :", i + 1)
    print("Total :", total[i])
    print("Pourcentage :", pourcentage[i], "%")
    print("Mention :", mention[i])

print("-------------------------")
```
